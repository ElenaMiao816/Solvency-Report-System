```python
import pandas as pd
import os
import glob
import camelot
%run CROSS_II_function.ipynb
# 假设 extract_tables_from_pdf 和 get_company_params 函数已定义
# from CROSS_II_function import extract_tables_from_pdf, get_company_params, offset

def process_pdf(pdf_file, company_name, fields, df_result, col_offset):
    print(company_name)
    
    keywords = [
    ['偿付能力', '寿险业务保险风险', '控制风险最低资本', '保险合同负债', '保单未来盈余'],
    ['认可资产', '认可负债', '实际资本']
    ]
    
    try:
        # 中邮的表格全有分隔线，使用pdfplumber抽表
        if company_name in ['中邮人寿','海保人寿','人保健康','华汇人寿']:
            extracted_tables = extract_tables_from_pdf(pdf_file)
        # 其他公司的表格不规则，使用camelot抽表
        else:
            extracted_tables = camelot.read_pdf(
                filepath=pdf_file,
                **get_company_params(company_name)
            )
        
        for field in fields:
            # 如果 df_result 中不存在 company_name 或 field，则初始化为 None
            if company_name not in df_result.index or field not in df_result.columns:
                df_result.at[company_name, field] = '无'
                
            found = False  # 添加一个标志变量
            for table in extracted_tables:
                value = offset(table.df if hasattr(table, 'df') else table, field, 0, col_offset)
        
                if value is None or value == "Index out of range" or value == "":
                    continue  # 如果值无效，跳过当前迭代

                for keyword_group in keywords:
                    if any(keyword in str(table.df.values.flatten()) if hasattr(table, 'df') else 
                           keyword in str(table.values.flatten()) for keyword in keyword_group):
                        if (df_result.at[company_name, field] is None or 
                            df_result.at[company_name, field] == '' or 
                            pd.isna(df_result.at[company_name, field]) or 
                            df_result.at[company_name, field] == '无'):
                            if '\n' in value:  #防止两列合并成一列
                                value = value.split('\n')[0]
                            df_result.at[company_name, field] = value
                            found = True  # 设置标志变量为 True
                            break  # 找到值后退出循环
                            
                if found:  # 如果找到值，退出 table 循环
                    break
                        
                       
    except Exception as e:
        print(f"处理文件 {pdf_file} 时发生错误: {e}")


fields = [
    #（一）主要偿付能力指标
    '综合偿付能力充足率', '核心偿付能力充足率', '综合偿付能力溢额',  '核心偿付能力溢额', 
    '认可资产', '认可负债','实际资本','核心一级资本', '核心二级资本','附属一级资本', '附属二级资本','最低资本',
    
    #（四）主要经营指标
    '保险业务收入','净利润','总资产','净资产', '保险合同负债', 
    
    # 实际资本表
    '计入核心一级资本的保单未来盈余', '计入核心二级资本的保单未来盈余',
    '计入附属一级资本的保单未来盈余', '计入附属二级资本的保单未来盈余',
    
    # 认可负债表
    '独立账户负债',

    # 最低资本
    '量化风险最低资本','量化风险最低资本（未考虑特征系数前）',
    
    '寿险业务保险风险最低资本合计', 
    '寿险业务保险风险-损失发生风险最低资本','寿险业务保险风险-退保风险最低资本',
    '寿险业务保险风险-费用风险最低资本','寿险业务保险风险-风险分散效应',
    
    '非寿险业务保险风险最低资本合计', 
    '非寿险业务保险风险-保费及准备金风险最低资本','非寿险业务保险风险-巨灾风险最低资本','非寿险业务保险风险-风险分散效应',
    
    '市场风险-最低资本合计',
    '市场风险-利率风险最低资本','市场风险-权益价格风险最低资本','市场风险-房地产价格风险最低资本',
    '市场风险-境外固定收益类资产价格风险最低资本','市场风险-境外权益类资产价格风险最低资本','市场风险-汇率风险最低资本','市场风险-风险分散效应',
    
    '信用风险-最低资本合计',
    '信用风险-利差风险最低资本','信用风险-交易对手违约风险最低资本','信用风险-风险分散效应',
    
    '量化风险分散效应', 
    '特定类别保险合同损失吸收效应', '损失吸收调整-不考虑上限','损失吸收效应调整上限','控制风险最低资本',
    
    '附加资本'
    '逆周期附加资本','D-SII附加资本','G-SII附加资本','其他附加资本'  
]

#fields = ['寿险业务保险风险-损失发生风险最低资本']

df_result = pd.DataFrame()

start_time = time.time()

# 指定目录路径
directory_path = r'D:\文档\_工作相关\KPMG\CROSS II\2025Q4\报告'

# 使用glob模块获取所有.pdf文件的路径
pdf_files = glob.glob(os.path.join(directory_path, '*.pdf'))

for pdf_file in pdf_files:
    company_name = pdf_file[len(directory_path)+1:pdf_file.rfind("20")]
    if company_name in ['财信人寿']: 
    #if company_name in ['中荷人寿']:
        if company_name in ['泰康养老']:  #泰康养老的格式特殊，先上期再本期
            process_pdf(pdf_file, company_name, fields, df_result, 2)
        else:
            process_pdf(pdf_file, company_name, fields, df_result, 1)
            process_pdf(pdf_file, company_name, fields, df_result, 2)

    

end_time = time.time()

elapsed_time = end_time - start_time
print(f"代码运行时间：{elapsed_time}秒")

# 将结果保存到 Excel 文件中
```
